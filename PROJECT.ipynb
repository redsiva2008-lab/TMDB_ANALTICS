```python
import pandas as pd
import numpy as np

movies = pd.read_csv("/content/sample_data/moviesoriginalindoc.csv")
genres = pd.read_csv("/content/sample_data/genres.csv")
movie_genres = pd.read_csv("/content/sample_data/movie_genres.csv")
cast = pd.read_csv("/content/sample_data/cast.csv")
crew = pd.read_csv("/content/sample_data/cast.csv")
movie_keywords = pd.read_csv("/content/sample_data/movie_keywords.csv")
for df in [movies, genres, movie_genres, cast, crew, movie_keywords]:
    df.columns = df.columns.astype(str).str.strip()
movies = movies.drop_duplicates(subset="movie_id", keep="first").copy()
for col in [
    "budget", "revenue", "runtime",
    "popularity", "vote_average", "vote_count"
]:
    movies[col] = pd.to_numeric(movies[col], errors="coerce")
movies["release_date"] = pd.to_datetime(
    movies["release_date"], errors="coerce"
)
movies["year"] = movies["release_date"].dt.year

print("Setup is done")
```

```python
print(movies.columns.tolist())
```

```python
numeric_cols = [
    "budget", "revenue", "runtime",
    "popularity", "vote_average", "vote_count"
]

missing = [col for col in numeric_cols if col not in movies.columns]
print("columns:", missing)
```

```python
movies.columns = movies.columns.str.strip()
```

```python
print("Unique movies:", movies["movie_id"].nunique())
print("Unique genres:", genres["genre_id"].nunique())
print("Unique cast members:", cast["person_id"].nunique())
print("Unique crew members:", crew["person_id"].nunique())
```

```python
finance_issues = movies[
    movies["budget"].isna()
    | movies["revenue"].isna()
    | (movies["budget"] <= 0)
    | (movies["revenue"] <= 0)
].copy()

finance_issues["issue"] = np.select(
    [
        finance_issues["budget"].isna() | (finance_issues["budget"] <= 0),
        finance_issues["revenue"].isna() | (finance_issues["revenue"] <= 0),
    ],
    ["Budget missing or zero", "Revenue missing or zero"],
    default="Budget and revenue missing or zero",
)

print("Keep these movies for other analyses; exclude them from profit/ROI.")
display(finance_issues[["movie_id", "title", "year", "budget", "revenue", "issue"]])
```

```python
valid_finance = movies[
    (movies["budget"] > 0) & (movies["revenue"] > 0)
].copy()

valid_finance["profit"] = valid_finance["revenue"] - valid_finance["budget"]
valid_finance["roi_percent"] = (
    valid_finance["profit"] / valid_finance["budget"] * 100
)

display(
    valid_finance[
        ["movie_id", "title", "year", "budget", "revenue", "profit", "roi_percent"]
    ].sort_values("title")
)
```

```python
top_15_roi = (
    valid_finance[valid_finance["budget"] >= 1_000_000]
    .nlargest(15, "roi_percent")
)

display(
    top_15_roi[
        ["title", "year", "budget", "revenue", "profit", "roi_percent"]
    ].reset_index(drop=True)
)
```

```python
movies_per_year = (
    movies.dropna(subset=["year"])
    .groupby("year")
    .agg(movie_count=("movie_id", "nunique"))
    .reset_index()
    .sort_values("year")
)

display(movies_per_year)

highest_year = movies_per_year.loc[movies_per_year["movie_count"].idxmax()]
print(
    f"Highest output: {int(highest_year['year'])} "
    f"({int(highest_year['movie_count'])} movies)"
)
```

```python
runtime = movies["runtime"].dropna()
q1 = runtime.quantile(0.25)
q3 = runtime.quantile(0.75)
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

runtime_outliers = movies[
    movies["runtime"].notna()
    & (
        (movies["runtime"] < lower_bound)
        | (movies["runtime"] > upper_bound)
    )
][["movie_id", "title", "year", "runtime"]].sort_values("runtime")

print(f"IQR outlier bounds: {lower_bound:g} to {upper_bound:g} minutes")
display(runtime_outliers)
```

```python
primary_genre = (
    movie_genres.drop_duplicates("movie_id", keep="first")
    .merge(genres, on="genre_id", how="left")
    .merge(
        movies[["movie_id", "title", "year", "popularity"]],
        on="movie_id",
        how="right",
    )
    .rename(columns={"genre_name": "primary_genre"})
)

display(
    primary_genre[
        ["movie_id", "title", "year", "primary_genre", "popularity"]
    ].sort_values("title")
)

genre_popularity = (
    primary_genre.groupby("primary_genre", dropna=False)
    .agg(
        movie_count=("movie_id", "nunique"),
        average_popularity=("popularity", "mean"),
    )
    .reset_index()
    .sort_values("average_popularity", ascending=False)
)

display(genre_popularity)
```

```python
cast_once_per_movie = cast.drop_duplicates(
    subset=["person_id", "movie_id"]
)

actor_ratings = cast_once_per_movie.merge(
    movies[["movie_id", "vote_average"]],
    on="movie_id",
    how="left",
)

top_10_actors = (
    actor_ratings.groupby(["person_id", "actor_name"])
    .agg(
        movie_count=("movie_id", "nunique"),
        average_movie_rating=("vote_average", "mean"),
    )
    .reset_index()
    .sort_values(
        ["movie_count", "average_movie_rating"],
        ascending=[False, False],
    )
    .head(10)
)

display(top_10_actors)
```

```python
repeated_titles = movies[
    movies["title"].notna()
    & movies.duplicated(subset="title", keep=False)
][["title", "year", "movie_id"]].sort_values(
    ["title", "year", "movie_id"]
)

display(repeated_titles)
```

```python
anomalies_data = movies.dropna(
    subset=["popularity", "vote_count"]
).copy()

anomalies_data = anomalies_data[
    (anomalies_data["popularity"] >= 0)
    & (anomalies_data["vote_count"] >= 0)
]

anomalies_data["log_votes"] = np.log1p(anomalies_data["vote_count"])
anomalies_data["log_popularity"] = np.log1p(anomalies_data["popularity"])

slope, intercept = np.polyfit(
    anomalies_data["log_votes"],
    anomalies_data["log_popularity"],
    1,
)

anomalies_data["residual"] = (
    anomalies_data["log_popularity"]
    - (slope * anomalies_data["log_votes"] + intercept)
)

anomalies_data["pattern"] = np.where(
    anomalies_data["residual"] >= 0,
    "High popularity compared with vote count",
    "Low popularity compared with vote count",
)

top_anomalies = anomalies_data.loc[
    anomalies_data["residual"].abs().nlargest(15).index,
    ["title", "year", "popularity", "vote_count", "residual", "pattern"],
].sort_values("residual")

display(top_anomalies)
```

```python
%%writefile app.py

import pandas as pd
import streamlit as st
from streamlit_option_menu import option_menu
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
st.set_page_config(layout='wide')
host= 'gateway01.ap-southeast-1.prod.aws.tidbcloud.com'
user = '3KvLe41uZryyAby.root'
passwd= 'HhQO8Gq1uutdvg8n'
database='test'
from sqlalchemy import create_engine
engine = create_engine(f'mysql+pymysql://{user}:{passwd}@{host}/{database}')

# Connect to sql


# Read all your cleaned dataframes
movies_df = pd.read_csv(r'/content/sample_data/moviesoriginalindoc.csv')
genre_df = pd.read_csv(r'/content/sample_data/geners.csv')
movie_genre_df=pd.read_csv(r"/content/sample_data/cleaned_movie_genres.csv")
crew_df= pd.read_csv(r"/content/sample_data/crew_cleaned.csv")
movie_keywoards_df=pd.read_csv(r"/content/sample_data/movie_keywoadsss.csv")
casts_df=pd.read_csv(r"/content/sample_data/cleaned_casts.csv")
st.title("TMDB Analytics")

with st.sidebar:
    selected = option_menu("Main Menu", ["Home",'Dashboard','Queries'],
        icons=['house'], menu_icon="cast", default_index=1)

if selected == "Home":
    c1,c2, c3 = st.columns([0.40,0.15,0.45])

    with c1:
                st.subheader("Hello welcome to my app",divider="rainbow")

if selected == 'Dashboard':

          movies_df["release_date"] = pd.to_datetime(movies_df["release_date"])

          movies_df["month"] = movies_df["release_date"].dt.strftime("%b")
          movies_df["day_of_month"] = movies_df["release_date"].dt.day

          fig = px.box(
          movies_df,
    x="month",
    y="day_of_month",
    points="all",
    title="Movie Release Day by Month",
    labels={
        "month": "Release Month",
        "day_of_month": "Day of Month"
    })
          st.plotly_chart(fig)

          genre_df=pd.read_csv("/content/sample_data/geners.csv")
          genre_df.title="genere"
          fig=px.density_heatmap(genre_df,x="genre_name",y="genre_id",title="genre")
          st.plotly_chart(fig)

          movie_genre_df = pd.read_csv(r"/content/sample_data/movie_genressss.csv")
          movie_genre_df["movie_id"] = pd.to_numeric(movie_genre_df["genre_id"], errors="coerce")

          fig = px.box(
          movie_genre_df,
    x="movie_id",
    y="genre_id",
    points="all",
    title="Movie Genre",
    labels={"movie_id": "Movie ID","genre_id": "genre_id"}
)
          st.plotly_chart(fig)

          genre_counts = (
          crew_df.groupby("department", as_index=False)
          .size()
          .rename(columns={"size": "crew_count"})
          .sort_values("crew_count", ascending=False)
)

          fig = px.line(
    genre_counts,
    x="department",
    y="crew_count",
    markers=True,
    title="Crew Records by Department",
    labels={
        "department": "Department",
        "crew_count": "Number of Crew Records"
    }
)

          fig.update_layout(xaxis_tickangle=-45)
          st.plotly_chart(fig)
          keyword_counts = (
          movie_keywoards_df.groupby("keyword_name")
          .size()
          .reset_index(name="movie_count")
          .sort_values("movie_count", ascending=False)
          .head(20)
)

          fig = px.bar(
    keyword_counts,
    x="keyword_name",
    y="movie_count",
    title="Top 20 Movie Keywords",
    labels={
        "keyword_name": "Movie Keyword",
        "movie_count": "Number of Movies"
    }
)

          fig.update_layout(xaxis_tickangle=-45)
          st.plotly_chart(fig)

          data = (
          casts_df.groupby("actor_name")["movie_id"]
          .nunique()
          .reset_index(name="movie_count")
          .sort_values("movie_count", ascending=False)
          .head(50)
)

          data["actor_rank"] = range(1, len(data) + 1)

          fig = px.scatter(
          data,
    x="actor_rank",
    y="movie_count",
    color="actor_name",
    size="movie_count",
    hover_data=["actor_name"],
    title="Movie Count by Actor",
    labels={
        "actor_rank": "Actor Rank",
        "movie_count": "Movie Count",
        "actor_name": "Actor"
    }
)

          st.plotly_chart(fig)

#3files upload
if selected == "Queries":

      option = st.selectbox(
      "Select a query",
      ("List all movies in the Action genre released after 2015, showing title and release date.",
      "Find the top 10 highest-grossing movies along with the genre(s) each belongs to",
      "Calculate the average budget and average revenue for each genre, sorted by average revenue descending.",
      "List the top 10 actors who have appeared in the most movies in the dataset.(cast)",
      "Find all directors who have directed more than 3 movies, along with their movie count.",
      "List the top 10 most frequently used keywords across all movies.",
      "Find movies where the budget was above the overall average budget but the revenue was below the overall average revenue.",
      "List all actors who have appeared in a movie directed by a specific director (e.g., Christopher Nolan",
      "Find the genre with the highest average rating, considering only genres with at least 20 movies.",
      "List the top 10 movies with the highest number of keywords tagged to them.")

  )
      if option == "List all movies in the Action genre released after 2015, showing title and release date.":

          df = pd.read_sql("""
SELECT DISTINCT m.title, m.release_date
FROM cleaned_orignal_movie_s AS m
JOIN movie_genressss AS mg ON m.movie_id = mg.movie_id
JOIN geners AS g ON mg.genre_id = g.genre_id
WHERE g.genre_name = 'Action'
  AND m.release_date >= '2016-01-01'
ORDER BY m.release_date;""",con=engine)
          st.dataframe(df)

          df["release_date"] = pd.to_datetime(df["release_date"])

          df["month"] = df["release_date"].dt.strftime("%b")
          df["day_of_month"] = df["release_date"].dt.day

          fig = px.box(
    df,
    x="month",
    y="day_of_month",
    points="all",
    title="Movie Release Day by Month",
    labels={
        "month": "Release Month",
        "day_of_month": "Day of Month"
    }
)

          st.plotly_chart(fig)

      if option == "Find the top 10 highest-grossing movies along with the genre(s) each belongs to":

          df=pd.read_sql("""
SELECT m.title, m.revenue,
       GROUP_CONCAT(DISTINCT g.genre_name ORDER BY g.genre_name SEPARATOR ', ') AS genres
FROM cleaned_orignal_movie_s AS m
LEFT JOIN movie_genressss AS mg ON m.movie_id = mg.movie_id
LEFT JOIN geners AS g ON mg.genre_id = g.genre_id
GROUP BY m.movie_id, m.title, m.revenue
ORDER BY m.revenue DESC
LIMIT 10;""",con=engine)
          st.dataframe(df)
          df["revenue"] = pd.to_numeric(df["revenue"], errors="coerce")
          df = df.dropna(subset=["revenue"]).sort_values("revenue", ascending=False)

          fig = px.line(
    df,
    x="title",
    y="revenue",
    markers=True,
    title="Movie Revenue",
    labels={
        "title": "Movie",
        "revenue": "Revenue"
    }
)

          fig.update_layout(xaxis_tickangle=-45)
          st.plotly_chart(fig)


      if option == "Calculate the average budget and average revenue for each genre, sorted by average revenue descending.":

          df = pd.read_sql("""
SELECT g.genre_name,
       AVG(m.budget) AS average_budget,
       AVG(m.revenue) AS average_revenue
FROM geners AS g
JOIN movie_genressss AS mg ON g.genre_id = mg.genre_id
JOIN cleaned_orignal_movie_s AS m ON mg.movie_id = m.movie_id
GROUP BY g.genre_id, g.genre_name
ORDER BY average_revenue DESC;""",con=engine)
          st.dataframe(df)

          fig = go.Figure(data=[
    go.Bar(
        x=df["genre_name"],
        y=df["average_budget"],
        name="Average Budget"
    ),
    go.Bar(
        x=df["genre_name"],
        y=df["average_revenue"],
        name="Average Revenue"
    )
])

          fig.update_layout(
    title="Average Movie Budget and Revenue by Genre",
    xaxis_title="Genre",
    yaxis_title="Average Amount",
    barmode="group",
    xaxis_tickangle=-45
)
          st.plotly_chart(fig)

      if option =="List the top 10 actors who have appeared in the most movies in the dataset.(cast)":

          df = pd.read_sql("""
SELECT actor_name, COUNT(DISTINCT movie_id) AS movie_count
FROM cleaned_casts
GROUP BY person_id, actor_name
ORDER BY movie_count DESC
LIMIT 10;""",con=engine)
          st.dataframe(df)

          df["movie_count"] = pd.to_numeric(df["movie_count"])


          fig = px.bar(
          df,
    x="actor_name",
    y="movie_count",
    text="movie_count",
    title="Movies by Actor",
    labels={
        "actor_name": "Actor",
        "movie_count": "Movie Count"
    }
)

          fig.update_layout(
    xaxis_tickangle=-45,
    xaxis_title="Actor",
    yaxis_title="Movie Count"
)
          st.plotly_chart(fig)
      if option =="Find all directors who have directed more than 3 movies, along with their movie count.":

          df = pd.read_sql("""SELECT person_name AS director_name,
       COUNT(DISTINCT movie_id) AS movie_count
FROM movies
WHERE job = 'Director'
GROUP BY person_id, person_name
HAVING COUNT(DISTINCT movie_id) > 3
ORDER BY movie_count DESC;""",con=engine)
          st.dataframe(df)
          df["director_rank"] = range(1, len(df) + 1)

          fig = px.scatter(
          df,
    x="director_rank",
    y="movie_count",
    hover_name="director_name",
    title="Movie Count by Director",
    labels={
        "director_rank": "Director Rank",
        "movie_count": "Movie Count"
    }
)
          st.plotly_chart(fig)
      if option =="List the top 10 most frequently used keywords across all movies.":

          df = pd.read_sql("""SELECT keyword_name, COUNT(DISTINCT movie_id) AS movie_count
FROM movie_keywoadsss
GROUP BY keyword_id, keyword_name
ORDER BY movie_count DESC
LIMIT 10;""",con=engine)
          st.dataframe(df)
          df["keyword_rank"] = range(1, len(df) + 1)

          fig = px.scatter(
          df,
    x="keyword_rank",
    y="movie_count",
    color="keyword_name",
    size="movie_count",
    hover_data=["keyword_name"],
    title="Movie Count by Keyword",
    labels={
        "keyword_rank": "Keyword Rank",
        "movie_count": "Movie Count",
        "keyword_name": "Keyword"
    }
)
          st.plotly_chart(fig)

      if option =="Find movies where the budget was above the overall average budget but the revenue was below the overall average revenue.":

          df = pd.read_sql("""SELECT title, budget, revenue
FROM cleaned_orignal_movie_s
WHERE budget > (SELECT AVG(budget) FROM cleaned_orignal_movie_s)
  AND revenue < (SELECT AVG(revenue) FROM cleaned_orignal_movie_s)
ORDER BY budget DESC;""",con=engine)
          st.dataframe(df)
          df["revenue"] = pd.to_numeric(df["revenue"], errors="coerce")
          df = df.dropna(subset=["revenue"]).sort_values("revenue", ascending=False)

          fig = px.funnel(
          df,
    x="revenue",
    y="title",
    title="Movie Revenue Funnel",
    labels={
        "revenue": "Revenue",
        "title": "Movie"
    }
)
          st.plotly_chart(fig)

      if option =="List all actors who have appeared in a movie directed by a specific director (e.g., Christopher Nolan":

          df = pd.read_sql("""SELECT DISTINCT c.actor_name
FROM cleaned_casts AS c
JOIN movies AS cr ON c.movie_id = cr.movie_id
WHERE cr.person_name = 'Christopher Nolan'
  AND cr.job = 'Director'
ORDER BY c.actor_name;""",con=engine)
          st.dataframe(df)

      if option =="Find the genre with the highest average rating, considering only genres with at least 20 movies.":

          df = pd.read_sql("""SELECT g.genre_name,
       AVG(m.vote_average) AS average_rating,
       COUNT(DISTINCT m.movie_id) AS movie_count
FROM geners AS g
JOIN movie_genressss AS mg ON g.genre_id = mg.genre_id
JOIN cleaned_orignal_movie_s AS m ON mg.movie_id = m.movie_id
GROUP BY g.genre_id, g.genre_name
HAVING COUNT(DISTINCT m.movie_id) >= 20
ORDER BY average_rating DESC
LIMIT 1;""",con=engine)
          st.dataframe(df)

      if option =="List the top 10 movies with the highest number of keywords tagged to them.":

          df = pd.read_sql("""SELECT m.title,
       COUNT(DISTINCT k.keyword_id) AS keyword_count
FROM cleaned_orignal_movie_s AS m
JOIN movie_keywoadsss AS k ON m.movie_id = k.movie_id
GROUP BY m.movie_id, m.title
ORDER BY keyword_count DESC
LIMIT 10;""",con=engine)
          st.dataframe(df)
          fig = go.Figure(data=[
    go.Bar(
        x=df["title"],
        y=df["keyword_count"]
    )
])

          fig.update_layout(
    title="Keyword Count by Movie",
    xaxis_title="Movie",
    yaxis_title="Keyword Count",
    xaxis_tickangle=-45
)
          st.plotly_chart(fig)
            # you need to display a chart along with table

```
